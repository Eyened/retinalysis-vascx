```python
from pathlib import Path

from matplotlib import pyplot as plt
from tqdm.notebook import tqdm

from vascx.fundus.features.od_morphology import ODMorphology
from vascx.fundus.loader import RetinaLoader
```

# Plot one feature on the samples

Change `FEATURE` below to plot another feature. Retina features produce one panel
per image; artery/vein features produce a panel for each available vessel layer.

```python
FEATURE = ODMorphology()

# Sample images, masks, and landmarks bundled with the repository.
SAMPLES_DIR = Path("../../samples/fundus")
# Save the complete sample grid alongside the other example figures.
OUTPUT_PATH = Path("../../samples/figures") / f"{FEATURE.canonical_name()}_samples.png"
N_COLS = 3
```

```python
loader = RetinaLoader.from_folder(SAMPLES_DIR)
panels = []
for i in tqdm(range(len(loader)), desc="Loading samples"):
    retina = loader[i]
    for target_name, target in retina.feature_targets(FEATURE):
        panels.append((retina.id, target_name, target))
```

```python
n_rows = (len(panels) + N_COLS - 1) // N_COLS
fig, axes = plt.subplots(
    n_rows, N_COLS, figsize=(6 * N_COLS, 6 * n_rows),
    squeeze=False, layout="constrained",
)
for ax, (image_id, target_name, target) in zip(axes.flat, panels):
    FEATURE.plot(ax, target, plot_caption=False)
    ax.set_title(f"{image_id} — {target_name}")
    ax.set_axis_off()
for ax in list(axes.flat)[len(panels):]:
    ax.set_axis_off()
plt.show()
```

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(OUTPUT_PATH, dpi=150, bbox_inches="tight")
print(OUTPUT_PATH.resolve())
```
