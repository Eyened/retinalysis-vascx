```python
from pathlib import Path

from vascx.fundus.loader import RetinaLoader
from vascx.utils.analysis import (
    extract_in_parallel,
)
from vascx.utils.feature_docs import write_biomarker_metadata
```

## Feature extraction

This notebook extracts biomarkers from the outputs of [1_segment_preprocessed.ipynb](1_segment_preprocessed.ipynb).

1. `RetinaLoader` loads the images, segmentations and preprocessing metadata.
2. VascX extracts biomarkers in parallel on the CPU.
3. The result has one row per image, with biomarker columns and `crop_scale_factor` metadata.
4. The final cell saves `biomarkers.csv` and its matching `data_dictionary.csv` together, ready for [3_post_process.ipynb](3_post_process.ipynb).

```python
# Folder containing the segmentations; both output CSV files are saved here.
ds_path = Path("../samples/fundus")
# Use the same feature set and naming for extraction and the data dictionary.
FEATURE_SET = "macula_centered"
NAMING = "resolved"
```

```python
loader = RetinaLoader.from_folder(ds_path)
```

```python
# Adjust n_jobs to your CPU. A starting point is half the available CPU threads.
res = extract_in_parallel(
    loader.to_dict(), FEATURE_SET, naming=NAMING, n_jobs=8, print_stack_trace=True
)
```

```python
res
```

```python
res.to_csv(ds_path / "biomarkers.csv", index_label="id")
write_biomarker_metadata(
    FEATURE_SET, ds_path / "data_dictionary.csv", naming=NAMING
)
```
